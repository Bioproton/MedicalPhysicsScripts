### Calculating coordinates

For each scintillator, each of these geometries need to be defined:
 
- Scintillator
- Outer casing box
- Inner casing box
- Shim cylinder (both pos and neg) 
- Shim1 hole pos
- Shim1 hole neg
- Shim2 hole pos
- Shim2 hole neg
- Cylinder pos [except for Layer 2]
- Cylinder neg [except for Layer 2]

Planes are shared (YZ- or XZ-plane) between scintillators within the same layer: 
- Casing pos
- Casing neg
- Shim1 pos
- Shim2 pos
- Shim 1 neg
- Shim 2 neg
